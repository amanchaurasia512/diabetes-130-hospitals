## 3. Initial Data Quality: Placeholder Values

The raw dataset uses `"?"` as a placeholder for missing or unavailable
information in several columns.

Before cleaning the data, we quantify these placeholder values to
understand the extent of missingness and identify columns that require
special attention during the data-cleaning and database-design stages.

```python
# Count how many '?' values exist in each column
missing_counts = (df == '?').sum()

# Show only columns that actually have at least one '?'
missing_counts = missing_counts[missing_counts > 0].sort_values(ascending=False)
print(missing_counts)
```

```python
question_mark_summary = pd.DataFrame({
    "missing_count": (df == "?").sum(),
    "missing_percentage": ((df == "?").sum() / len(df)) * 100
})

question_mark_summary = (
    question_mark_summary[question_mark_summary["missing_count"] > 0]
    .sort_values("missing_count", ascending=False)
)

question_mark_summary
```

```python
column_profile = pd.DataFrame({
    "data_type": df.dtypes,
    "unique_values": df.nunique(),
    "missing_values": df.isna().sum(),
    "question_marks": (df == "?").sum()
})

column_profile
```

### Initial Observations

The raw dataset contains 101,766 encounter records and 71,518 unique
patient identifiers. The encounter identifier is unique across the

require further investigation as part of the eventual database design.

Some variables have very low or no variation, including `examide` and

exclusion decisions.

### Patient–Encounter Relationship

The dataset contains 101,766 encounters associated with 71,518 unique
patient identifiers. This indicates that the source data is encounter-
level rather than patient-level.

Most patients have a single recorded encounter, while a subset has
multiple encounters. The maximum observed number of encounters for a
single patient identifier is 40.

This supports a one-to-many relationship between patient identifiers
and encounters for the purpose of database design. However, the repeated
encounters should not be interpreted as a complete longitudinal patient
history because the source dataset does not provide sufficient
information to reconstruct one.

```python
patient_consistency = df.groupby("patient_nbr").agg(
    race_unique=("race", "nunique"),
    gender_unique=("gender", "nunique"),
    age_unique=("age", "nunique")
)

patient_consistency.describe()
```

```python
patient_consistency[
    (patient_consistency["race_unique"] > 1) |
    (patient_consistency["gender_unique"] > 1) |
    (patient_consistency["age_unique"] > 1)
].head(20)
```

```python
attribute_consistency_summary = {
    "Race varies": (patient_consistency["race_unique"] > 1).sum(),
    "Gender varies": (patient_consistency["gender_unique"] > 1).sum(),
    "Age varies": (patient_consistency["age_unique"] > 1).sum()
}

attribute_consistency_summary
```

```python
attribute_consistency_pct = {
    "Race varies (%)": (patient_consistency["race_unique"] > 1).mean() * 100,
    "Gender varies (%)": (patient_consistency["gender_unique"] > 1).mean() * 100,
    "Age varies (%)": (patient_consistency["age_unique"] > 1).mean() * 100
}

attribute_consistency_pct
```

### Patient Attribute Consistency

Among the 71,518 unique patient identifiers, race is consistent across
repeated encounters for almost all patients. Only 249 patients (0.35%)
have more than one observed race value.

Gender is highly consistent, with only 3 patients (0.004%) showing more
than one observed gender value.

Age band shows more variation: 1,574 patients (2.20%) have more than one
observed age band across their encounters. Unlike race and gender, age
band can naturally change between encounters as patients age.

These results show that repeated encounter records generally contain
consistent demographic information, while age band can vary over time.
No values are modified at this profiling stage.
